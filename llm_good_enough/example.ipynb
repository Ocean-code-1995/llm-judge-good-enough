#### ***Read Data***

```python
import pandas as pd

df = pd.read_csv('../wmt-human/data/wmt-clean.csv')
print(df.shape)
df.head()
```

```python
from evaluator import LLMGoodEnough

# define human judge columns
human_cols = [col for col in df.columns if col.startswith('human_')]

# initialize evaluator instance
LLM_Evaluator = LLMGoodEnough(df=df, human_cols=human_cols, min_score=0, max_score=6)

# visualize LLM-as-a-judge good enough
LLM_Evaluator.visulize_good_enough(
    llm_col='LLAMA_as_a_judge',        # LLM-as-a-judge column name
    y_lim=0.6,                         # y-axis limit -> probability (0-1)
)
```
